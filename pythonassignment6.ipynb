```python
import numpy as np
import pandas as pd
raw_data={"Roll_no.":[101, 102, 103, 104, 105], "Name":["Amit","Rahul","Ira","Sania","Karan"],"Marks":[85.0, np.nan, np.nan, 90.0, np.nan], "City":["Pune","Mumbai","Kolkata","Delhi","Pune"]}
df=pd.DataFrame(raw_data)
#display the original dataset
print("Original DataFrame:")
print(df)
#display the first five rows
print("\nFirst five rows:")
print(df.head())
#display dataset information
print("\nDataFrame information:")
print(df.info())
#display statistical summary
print("\nStatistical summary:")
print(df.describe())
#check for missing values
print("nMissing values:")
print(df.isnull().sum())
#Remove rows containing missing values
print("\nData after dropna():")
print(df.dropna())
#Fill missing marks with the mean
df["Marks"]=df["Marks"].fillna(df["Marks"].mean())
print("\nData After Filling Missing Marks:")
print(df)

# Identify duplicate rows
print("\nDuplicate Rows:")
print(df.duplicated())

# Remove duplicate rows
df = df.drop_duplicates()

print("\nData After Removing Duplicates:")
print(df)

# Drop an irrelevant column (example)
df_without_city = df.drop(columns=["City"])

print("\nData After Dropping City Column:")
print(df_without_city)

# Select specific columns using loc[]
print("\nSelected Student Data:")
print(df.loc[:, ["Name", "Marks"]])

# Filter students who scored more than 85 marks
print("\nStudents Scoring More Than 85 Marks:")
print(df.loc[df["Marks"] > 85])

# Display the final cleaned dataset
print("\nFinal Cleaned DataFrame:")
print(df)
```
